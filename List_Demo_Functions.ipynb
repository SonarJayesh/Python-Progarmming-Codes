```python
# WAP to create a list containing first 100 numbers as the elements
num_list = [] 
for i in range(0,100):
    num_list.append(i)
print(num_list)    
```

```python
# WAP to create list that contains odd numbers(less than 200)
odd_list = []
for i in range(0,200):
    if i % 2 == 0: 
        odd_list.append(i)
print(odd_list)    
```
